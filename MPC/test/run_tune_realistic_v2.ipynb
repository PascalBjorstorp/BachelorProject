# Run `tune_realistic_v2.py`

This notebook launches the MPC tuner script from the local `MPC` folder and streams live logs.

Steps:
1. Run the setup cell.
2. Edit arguments in the config cell.
3. Run the execution cell.
4. Inspect the latest CSV files.

```python
from pathlib import Path
import subprocess
import sys
import os

REPO_ROOT = Path.cwd().resolve().parent.parent  # assumes notebook under MPC/test
MPC_DIR = REPO_ROOT / "MPC"
SCRIPT_PATH = MPC_DIR / "test" / "tune_realistic_v2.py"

print("REPO_ROOT:", REPO_ROOT)
print("MPC_DIR:", MPC_DIR)
print("SCRIPT_PATH exists:", SCRIPT_PATH.exists())
assert SCRIPT_PATH.exists(), f"Missing script: {SCRIPT_PATH}"
```

```python
# Edit these args as needed.
RUN_ARGS = [
    "--jobs", "1",
    "--phase2-top", "1",
    "--global-passes", "1",
    "--no-obstacles",
]

COMMAND = [sys.executable, str(SCRIPT_PATH)] + RUN_ARGS
print("Command:", " ".join(COMMAND))
```

```python
# Run and stream logs live.
process = subprocess.Popen(
    COMMAND,
    cwd=str(MPC_DIR),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
)

try:
    assert process.stdout is not None
    for line in process.stdout:
        print(line, end="")
finally:
    rc = process.wait()

print("\nReturn code:", rc)
```

```python
# Show latest output files.
out_dir = MPC_DIR / "test"
latest_csv = sorted(out_dir.glob("tuning_hardware_base_*.csv"), key=lambda p: p.stat().st_mtime, reverse=True)[:5]
latest_sorted = sorted(out_dir.glob("tuning_hardware_base_*_sorted.csv"), key=lambda p: p.stat().st_mtime, reverse=True)[:5]

print("Latest CSV files:")
for p in latest_csv:
    print(" ", p.name)

print("\nLatest sorted CSV files:")
for p in latest_sorted:
    print(" ", p.name)
```
